```python
from pathlib import Path
import pdfplumber
import pandas as pd

# Find the project root whether Jupyter starts from the repo root
# or from inside the notebooks folder.
cwd = Path.cwd()

possible_paths = [
    cwd / "source_docs" / "2023-24_sa_fee_memo.pdf",
    cwd.parent / "source_docs" / "2023-24_sa_fee_memo.pdf",
]

pdf_path = next(
    (path.resolve() for path in possible_paths if path.exists()),
    None
)

print("Current working directory:", cwd)
print("PDF found:", pdf_path)
```

```python
with pdfplumber.open(pdf_path) as pdf:
    print("Number of pages:", len(pdf.pages))

    for page_number, page in enumerate(pdf.pages, start=1):
        text = page.extract_text()

        print("\n" + "=" * 70)
        print(f"PAGE {page_number}")
        print("=" * 70)

        if text:
            print(text[:1500])
        else:
            print("No readable text detected.")
```

```python
with pdfplumber.open(pdf_path) as pdf:
    for page_number, page in enumerate(pdf.pages, start=1):
        tables = page.extract_tables()

        print(
            f"Page {page_number}: "
            f"{len(tables)} table(s) detected"
        )

        for table_number, table in enumerate(tables, start=1):
            print(f"\nTable {table_number}")
            for row in table[:10]:
                print(row)
```

```python
with pdfplumber.open(pdf_path) as pdf:
    page = pdf.pages[2]  # Page 3
    table = page.extract_tables()[0]

# Skip the first row because it is the merged PDF header.
budget_rows = table[1:]

df_2023_24_raw = pd.DataFrame(
    budget_rows,
    columns=[
        "program_name_raw",
        "allocation_2022_23_raw",
        "request_2023_24_raw",
        "allocation_2023_24_raw",
    ],
)

display(df_2023_24_raw)

print("\nRows extracted:", len(df_2023_24_raw))
```

```python
print("First 5 rows:")
display(df_2023_24_raw.head())

print("Last 5 rows:")
display(df_2023_24_raw.tail())
```

```python
def parse_currency_for_check(value):
    if value is None:
        return 0.0

    value = str(value).strip()

    if value in {"", "-", "—"}:
        return 0.0

    value = (
        value
        .replace("$", "")
        .replace(",", "")
    )

    return float(value)


validation_totals = {
    "2022-23 Allocation": df_2023_24_raw[
        "allocation_2022_23_raw"
    ].apply(parse_currency_for_check).sum(),

    "2023-24 Request": df_2023_24_raw[
        "request_2023_24_raw"
    ].apply(parse_currency_for_check).sum(),

    "2023-24 Allocation": df_2023_24_raw[
        "allocation_2023_24_raw"
    ].apply(parse_currency_for_check).sum(),
}

for label, total in validation_totals.items():
    print(f"{label}: ${total:,.2f}")
```

```python
# Create a traceable raw copy
df_2023_24_export = df_2023_24_raw.copy()

df_2023_24_export.insert(
    0,
    "source_document",
    "2023-24_sa_fee_memo.pdf"
)

df_2023_24_export.insert(
    1,
    "source_page",
    3
)

df_2023_24_export.insert(
    2,
    "source_url",
    "https://studentleadership.seattlecentral.edu/sites/"
    "studentleadership.seattlecentral.edu/files/"
    "2023-09/2223-SA-Fees-Memo.pdf"
)

display(df_2023_24_export.head())
```

```python
with pdfplumber.open(pdf_2024_25) as pdf:
    page = pdf.pages[2]  # Page 3
    table = page.extract_tables()[0]

# First two rows are the multi-row table header
budget_rows_2024_25 = table[2:]

df_2024_25_raw = pd.DataFrame(
    budget_rows_2024_25,
    columns=[
        "program_name_raw",
        "allocation_2023_24_raw",
        "request_2024_25_raw",
        "allocation_2024_25_raw",
    ],
)

display(df_2024_25_raw)

print("\nRows extracted:", len(df_2024_25_raw))
```

```python
# Remove the completely blank extraction row
df_2024_25_raw = df_2024_25_raw[
    df_2024_25_raw["program_name_raw"].notna()
].copy()

df_2024_25_raw["program_name_raw"] = (
    df_2024_25_raw["program_name_raw"].astype(str).str.strip()
)

df_2024_25_raw = df_2024_25_raw[
    df_2024_25_raw["program_name_raw"] != ""
].copy()

# Preserve the reported Total row separately
reported_total_2024_25 = df_2024_25_raw[
    df_2024_25_raw["program_name_raw"].str.lower() == "total"
].copy()

# Program-level rows only
df_2024_25_programs = df_2024_25_raw[
    df_2024_25_raw["program_name_raw"].str.lower() != "total"
].copy()

print("Program rows:", len(df_2024_25_programs))

display(reported_total_2024_25)
```

```python
calculated_totals_2024_25 = {
    "2023-24 Allocation": df_2024_25_programs[
        "allocation_2023_24_raw"
    ].apply(parse_currency_for_check).sum(),

    "2024-25 Request": df_2024_25_programs[
        "request_2024_25_raw"
    ].apply(parse_currency_for_check).sum(),

    "2024-25 Allocation": df_2024_25_programs[
        "allocation_2024_25_raw"
    ].apply(parse_currency_for_check).sum(),
}

reported_totals_2024_25 = {
    "2023-24 Allocation": parse_currency_for_check(
        reported_total_2024_25.iloc[0]["allocation_2023_24_raw"]
    ),

    "2024-25 Request": parse_currency_for_check(
        reported_total_2024_25.iloc[0]["request_2024_25_raw"]
    ),

    "2024-25 Allocation": parse_currency_for_check(
        reported_total_2024_25.iloc[0]["allocation_2024_25_raw"]
    ),
}

for label in calculated_totals_2024_25:
    calculated = calculated_totals_2024_25[label]
    reported = reported_totals_2024_25[label]

    print(f"{label}")
    print(f"  Calculated: ${calculated:,.2f}")
    print(f"  Reported:   ${reported:,.2f}")
    print(f"  Difference: ${calculated - reported:,.2f}")
    print()
```

```python
rows_2025_26 = []

with pdfplumber.open(pdf_2025_26) as pdf:

    # PAGE 1
    table_page_1 = pdf.pages[0].extract_tables()[0]

    # Skip:
    # row 0 = blank
    # row 1 = fiscal-year header
    # row 2 = Allocation / Request header
    for row in table_page_1[3:]:
        rows_2025_26.append(
            [1] + row
        )

    # PAGE 2
    table_page_2 = pdf.pages[1].extract_tables()[0]

    # Page 2 is a continuation, so there is no header to skip
    for row in table_page_2:
        rows_2025_26.append(
            [2] + row
        )


df_2025_26_raw = pd.DataFrame(
    rows_2025_26,
    columns=[
        "source_page",
        "program_name_raw",
        "allocation_2023_24_raw",
        "allocation_2024_25_raw",
        "request_2025_26_raw",
        "allocation_2025_26_raw",
    ],
)

display(df_2025_26_raw)

print("\nRows extracted:", len(df_2025_26_raw))
```
